```python
import matplotlib.pyplot as plt

# Delivery count by destination zone
zone_counts = logistics_df["Destination_Zone"].value_counts()

plt.figure(figsize=(9, 5))
zone_counts.plot(kind="bar")
plt.title("Number of Deliveries by Destination Zone")
plt.xlabel("Destination Zone")
plt.ylabel("Number of Deliveries")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Average delivery time by destination zone

avg_time_zone = (
    logistics_df.groupby("Destination_Zone")["Delivery_Time_Minutes"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
avg_time_zone.plot(kind="bar")
plt.title("Average Delivery Time by Destination Zone")
plt.xlabel("Destination Zone")
plt.ylabel("Average Delivery Time (Minutes)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Average delivery cost by destination zone

avg_cost_zone = (
    logistics_df.groupby("Destination_Zone")["Delivery_Cost_INR"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
avg_cost_zone.plot(kind="bar")
plt.title("Average Delivery Cost by Destination Zone")
plt.xlabel("Destination Zone")
plt.ylabel("Average Delivery Cost (INR)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
traffic_analysis = logistics_df.groupby("Traffic_Level").agg(
    Deliveries=("Delivery_ID", "count"),
    Avg_Delivery_Time_Minutes=("Delivery_Time_Minutes", "mean"),
    Avg_Delivery_Cost_INR=("Delivery_Cost_INR", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).round(2)

traffic_analysis
```

```python
traffic_time = (
    logistics_df.groupby("Traffic_Level")["Delivery_Time_Minutes"]
    .mean()
    .reindex(["Low", "Medium", "High"])
)

plt.figure(figsize=(8, 5))
traffic_time.plot(kind="bar")
plt.title("Average Delivery Time by Traffic Level")
plt.xlabel("Traffic Level")
plt.ylabel("Average Delivery Time (Minutes)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
weather_analysis = logistics_df.groupby("Weather_Condition").agg(
    Deliveries=("Delivery_ID", "count"),
    Avg_Delivery_Time_Minutes=("Delivery_Time_Minutes", "mean"),
    Avg_Delivery_Cost_INR=("Delivery_Cost_INR", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).round(2)

weather_analysis
```

```python
weather_time = (
    logistics_df.groupby("Weather_Condition")["Delivery_Time_Minutes"]
    .mean()
    .reindex(["Normal", "Rain", "Storm"])
)

plt.figure(figsize=(8, 5))
weather_time.plot(kind="bar")
plt.title("Average Delivery Time by Weather Condition")
plt.xlabel("Weather Condition")
plt.ylabel("Average Delivery Time (Minutes)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    logistics_df["Distance_KM"],
    logistics_df["Delivery_Cost_INR"],
    alpha=0.6
)

plt.title("Relationship Between Delivery Distance and Cost")
plt.xlabel("Distance (KM)")
plt.ylabel("Delivery Cost (INR)")
plt.tight_layout()
plt.show()
```

```python
distance_cost_correlation = (
    logistics_df["Distance_KM"]
    .corr(logistics_df["Delivery_Cost_INR"])
)

print(
    f"Correlation between distance and delivery cost: "
    f"{distance_cost_correlation:.3f}"
)
```

## 7. KPI Analysis

```python
# ============================================================
# KEY PERFORMANCE INDICATORS (KPIs)
# ============================================================

# 1. On-Time Delivery Rate
on_time_rate = (
    (logistics_df["On_Time_Delivery"] == "Yes").mean() * 100
)

# 2. Average Delivery Time
average_delivery_time = logistics_df["Delivery_Time_Minutes"].mean()

# 3. Average Delivery Cost
average_delivery_cost = logistics_df["Delivery_Cost_INR"].mean()

# 4. Average Vehicle Utilization
logistics_df["Vehicle_Utilization_%"] = (
    logistics_df["Load_Weight_KG"]
    / logistics_df["Vehicle_Capacity_KG"]
    * 100
)

average_vehicle_utilization = (
    logistics_df["Vehicle_Utilization_%"].mean()
)

# Display KPI results
print("LOGISTICS KPI SUMMARY")
print("=" * 35)

print(f"On-Time Delivery Rate: {on_time_rate:.2f}%")
print(f"Average Delivery Time: {average_delivery_time:.2f} minutes")
print(f"Average Delivery Cost: ₹{average_delivery_cost:.2f}")
print(f"Average Vehicle Utilization: {average_vehicle_utilization:.2f}%")
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "On-Time Delivery Rate",
        "Average Delivery Time",
        "Average Delivery Cost",
        "Average Vehicle Utilization"
    ],
    "Value": [
        f"{on_time_rate:.2f}%",
        f"{average_delivery_time:.2f} minutes",
        f"₹{average_delivery_cost:.2f}",
        f"{average_vehicle_utilization:.2f}%"
    ]
})

kpi_summary
```

```python
vehicle_kpi = logistics_df.groupby("Vehicle_Type").agg(
    Deliveries=("Delivery_ID", "count"),
    Avg_Delivery_Time_Minutes=("Delivery_Time_Minutes", "mean"),
    Avg_Delivery_Cost_INR=("Delivery_Cost_INR", "mean"),
    Avg_Utilization_Percent=("Vehicle_Utilization_%", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).round(2)

vehicle_kpi
```

```python
zone_kpi = logistics_df.groupby("Destination_Zone").agg(
    Deliveries=("Delivery_ID", "count"),
    Avg_Distance_KM=("Distance_KM", "mean"),
    Avg_Delivery_Time_Minutes=("Delivery_Time_Minutes", "mean"),
    Avg_Delivery_Cost_INR=("Delivery_Cost_INR", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).round(2)

zone_kpi
```
